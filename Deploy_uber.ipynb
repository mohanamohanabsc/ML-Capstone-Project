```python
import pickle
import pandas as pd
```

```python
pickup_long=float(input("Enter the pickup_longitude:"))
pickup_la=float(input("Enter the pickup_latitude:"))
dropoff_long=float(input("Enter the dropoff_longitude:"))
dropoff_la=float(input("Enter the dropoff_latitude:"))
year=int(input("Enter the year:"))
```

```python
loaded_model_d=pickle.load(open("Final_model_uber_dist.sav",'rb'))
loaded_model_f=pickle.load(open("Final_model_uber_fare.sav",'rb'))
dist_pred=loaded_model_d.predict([[pickup_long, pickup_la,dropoff_long, dropoff_la, year]])[0]
fare_pred=loaded_model_f.predict([[pickup_long, pickup_la,dropoff_long, dropoff_la, year]])[0]
dist_pred_km = dist_pred / 1000
print("Predicted Distance:",round(dist_pred_km, 2), "km")
print("Predicted Uber Fare: ₹",round(fare_pred, 2))
```
